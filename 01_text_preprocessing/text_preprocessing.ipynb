# Removing HTML

## remove all tags and retain textural information 

```python
sample_for_html_remove = csv_data['review'][1]
print(sample_for_html_remove)
```

```python
import re

def rmove_html_tag(text):
    pattern = re.compile('<.*?>')
    replaced = pattern.sub(r'', text)
    return replaced
    

html_removed = rmove_html_tag(sample_for_html_remove)
print(html_removed)
```

```python
# removig html for entire corpus

csv_data['review']=csv_data['review'].apply(rmove_html_tag)
print(csv_data)
```

```python
# remove url
import re

def remove_url(text):
    pattern= re.compile(r'https?://\S+|www\.\S+')  #re.compile(r'https?\s*://\s*\S+|www\.\s*\S+')
    removed_url_text = pattern.sub(r'', text)
    return removed_url_text

same_text = "it is www.kaggle.com"

url_pre= remove_url(same_text)
print(url_pre)
```
